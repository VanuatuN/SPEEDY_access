```python
import xarray as xr
import matplotlib.pyplot as plt

f = "/leonardo_work/ICT26_ESP/ntilinin/ERA5/era5_sw_1958.nc"

ds = xr.open_dataset(f)
print(ds)
print(ds.data_vars)

v = "avg_sdswrf"
t = "valid_time" if "valid_time" in ds.coords else "time"

x = ds[v]

jjas = x.where(x[t].dt.month.isin([6,7,8,9]), drop=True).mean(t).mean("longitude")
djfm = x.where(x[t].dt.month.isin([12,1,2,3]), drop=True).mean(t).mean("longitude")

fig, ax = plt.subplots(figsize=(10,5))

ax.plot(jjas.latitude, jjas, lw=2.5, label="ERA5 JJAS")
ax.plot(djfm.latitude, djfm, lw=2.5, label="ERA5 DJFM")

ax.set_xlim(-90,90)
ax.set_xlabel("Latitude")
ax.set_ylabel("Net surface solar radiation [W m$^{-2}$]")
ax.set_title("ERA5 zonal-mean net surface solar radiation, 1958")
ax.grid(alpha=.25)
ax.legend(frameon=False)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```
